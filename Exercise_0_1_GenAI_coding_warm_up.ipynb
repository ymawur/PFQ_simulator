## Dataframe

## Plotting the data

Once the data are in the Python environment (e.g. Kiwi_data), we can start to make graphs of data. Here, we again can prompt Gen AI to write the codes for us.

Here a simple plot can be made to outline the Kiwi firmness change overtime.

### Prompt:

**Plot the Time vs Firmness graph of a dataframe called "Kiwi_data". Color code the lines by values from the Temperature column in python**

Note: here, we need to prompt the exact names of the column names (Temperature,	Time,	Firmness) to provide enough information for the AI to write the code.


````python
import matplotlib.pyplot as plt
import pandas as pd

# assuming Kiwi_data is already a pandas DataFrame
# with columns: Time, Firmness, Temperature

fig, ax = plt.subplots()

# group by temperature so each temperature gets its own line
for temp, df in Kiwi_data.groupby("Temperature"):
    ax.scatter(df["Time"], df["Firmness"], label=f"{temp}")

ax.set_xlabel("Time")
ax.set_ylabel("Firmness")
ax.set_title("Time vs Firmness by Temperature")
ax.legend(title="Temperature")
plt.show()
````

## Function writing

A function is a reusable block of code that takes inputs, performs some computation, and returns an output.

- Inputs (arguments) are values you pass into the function so it can do its job.

- The function body contains the operations or logic you want to run.

- The output (return value) is what the function gives back after it finishes the computation.

For example, we can turn the linear regression equation $y = ax + b$ in to a function by specifying the inputs and outputs of the equation.

Run the code below, you can see that we first need to define a function based on the mathematical relation. After that, in the demo section, we can change inputs (x, a, b) to calculate the output (y).

Feel free to change the values in the inputs, x, a, b, and run the code again.

````python
def linear_model(x, a, b):
    """Return y = a*x + b."""
    return a * x + b

# demo
x = 5
a = 2
b = 3
y = linear_model(x, a, b)
print("y =", y)
````

You can also implement function into the calculating new columns in a dataframe. For example, in the Kiwi_data, we have data about firmness. Also, we know that measured firmness has a inverse linear relationship with ripeness of the kiwi fruit. The relation is outline below.

$
\text{Ripeness}(F) =
\begin{cases}
1, & F < 3, \\[6pt]
1 - \dfrac{F - 3}{20 - 3}, & 3 \le F \le 20.
\end{cases}
$

For firmness values between 3 and 20, ripeness decreases in a straight line as firmness increases. At a firmness of 3 the fruit is still fully ripe, and as firmness rises toward 20 the fruit becomes progressively less ripe, reaching the minimum ripeness at the highest firmness.

Of course, we can also prompt an AI to make this into a function describing the relation between firmness and ripeness.

###Prompt

```
\text{Ripeness}(F) =
\begin{cases}
1, & F < 3, \\[6pt]
1 - \dfrac{F - 3}{20 - 3}, & 3 \le F \le 20.
\end{cases}
```


**Based on this relation, prepare a function with an input of Firmness and an output of Ripeness in python**



````python
def ripeness(F):
    """
    Compute ripeness based on firmness F using the rule:
      - Ripeness = 1 when F < 3
      - Ripeness decreases linearly from 1 at F = 3 to 0 at F = 20
    """
    if F < 3:
        return 1.0

    # linear inverse mapping from F in [3, 20] to ripeness in [1, 0]
    return 1 - (F - 3) / (20 - 3)

# Example

ripeness(10)
````

Based on this function, we can then create a new column in the Kiwi_data dataframe to calculate the ripeness for fruits stored at different time and temperature.

###Prompt

Use the function `ripeness`, add a column in Kiwi_data as "Ripeness" by using the "Firmness" column as input.

````python
# add Ripeness column based on the Firmness column
Kiwi_data["Ripeness"] = Kiwi_data["Firmness"].apply(ripeness)

Kiwi_data
````

Of course, we can also plot the time and temperature effects on ripeness now.

###Prompt:

**Plot the Time vs Ripeness in the Kiwi_data and color coded them by Temperature**

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

# group and color by Temperature
temps = Kiwi_data["Temperature"].unique()
cmap = plt.cm.get_cmap("viridis", len(temps))

for i, temp in enumerate(temps):
    df_temp = Kiwi_data[Kiwi_data["Temperature"] == temp]
    ax.scatter(df_temp["Time"], df_temp["Ripeness"],
               color=cmap(i),
               label=f"{temp}")

ax.set_xlabel("Time", fontsize=14)
ax.set_ylabel("Ripeness", fontsize=14)
ax.set_title("Time vs Ripeness Color Coded by Temperature", fontsize=16)

ax.legend(title="Temperature", fontsize=12, title_fontsize=12)
plt.show()
````
